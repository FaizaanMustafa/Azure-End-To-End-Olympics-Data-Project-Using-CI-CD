```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from delta.tables import *
```

```python
dbutils.widgets.text('source_container_name','')
dbutils.widgets.text('source_folder','')
dbutils.widgets.text('source_folder_directory','')
dbutils.widgets.text('target_folder_directory','')
dbutils.widgets.text('target_folder','')
source_container_name = dbutils.widgets.get('source_container_name')
source_folder = dbutils.widgets.get('source_folder')
source_folder_directory = dbutils.widgets.get('source_folder_directory')
target_folder = dbutils.widgets.get('target_folder')
target_folder_directory = dbutils.widgets.get('target_folder_directory')
```

```python
df = spark.read.format('csv').option('header','true').option('inferSchema','true').load(f'abfss://{source_container_name}@adls7428.dfs.core.windows.net/{source_folder}/{source_folder_directory}')
df.display()
```

```python
df.write.format('delta').mode('append').option('path',f'abfss://{source_container_name}@adls7428.dfs.core.windows.net/{target_folder}/{target_folder_directory}').saveAsTable(f'olympics.silver.silver_{target_folder_directory}')
```
