```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = spark.read.format('csv').option('header','true').option('inferschema','true').load('abfss://olympics@adls7428.dfs.core.windows.net/bronze/nocs')
df.display()
```

```python
df = df.withColumn('tag',split(col('tag'),'-')[0])
display(df)
```

```python
df.write.format('delta').mode('append').option('path','abfss://olympics@adls7428.dfs.core.windows.net/silver/nocs').saveAsTable('olympics.silver.nocs')
```
